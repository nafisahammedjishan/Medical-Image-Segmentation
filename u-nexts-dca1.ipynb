```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# Single Scale Retinex
# ==========================================================
def single_scale_retinex(img, sigma=30):

    img = img.astype(np.float32) + 1

    blur = cv2.GaussianBlur(img, (0,0), sigma)

    retinex = np.log(img) - np.log(blur + 1)

    retinex = cv2.normalize(
        retinex,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return retinex.astype(np.uint8)


# ==========================================================
# Multi-scale TopHat
# ==========================================================
def multiscale_tophat(img):

    result = img.astype(np.int16)

    for s in [5,9,13]:

        kernel = cv2.getStructuringElement(
            cv2.MORPH_ELLIPSE,
            (s,s)
        )

        white = cv2.morphologyEx(
            img,
            cv2.MORPH_TOPHAT,
            kernel
        )

        black = cv2.morphologyEx(
            img,
            cv2.MORPH_BLACKHAT,
            kernel
        )

        result += white
        result -= black

    result = np.clip(result,0,255)

    return result.astype(np.uint8)


# ==========================================================
# Guided Filter
# ==========================================================
def guided(img):

    return cv2.ximgproc.guidedFilter(
        guide=img,
        src=img,
        radius=7,
        eps=1e-3
    )


from skimage.filters import frangi

def frangi_filter(img):

    vessel = frangi(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)



def gamma_correction(img, gamma=0.8):

    table = np.array([
        ((i / 255.0) ** gamma) * 255
        for i in np.arange(256)
    ]).astype("uint8")

    return cv2.LUT(img, table)



from skimage.filters import sato

def sato_filter(img):

    vessel = sato(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)


from skimage.filters import hessian

def hessian_filter(img):

    vessel = hessian(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)


def edge_preserving(img):

    img_bgr = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

    out = cv2.edgePreservingFilter(
        img_bgr,
        flags=1,
        sigma_s=60,
        sigma_r=0.4
    )

    return cv2.cvtColor(out, cv2.COLOR_BGR2GRAY)


# ==========================================================
# Complete Pipeline
# ==========================================================
def preprocess_xca(img):

    # CLAHE
    clahe = cv2.createCLAHE(
        clipLimit=2,
        tileGridSize=(8,8)
    )

    img1 = clahe.apply(img)

    # Circular Gaussian
    img2 = cv2.GaussianBlur(
        img1,
        (7,7),
        1.5
    )

    # Guided Filter
    img3 = guided(img2)

    # Multi-scale TopHat
    img4 = multiscale_tophat(img3)

    # edge preserving
    img5 = edge_preserving(
        img4,
    )

    # Retinex instead of Homomorphic
    img6 = single_scale_retinex(
        img5,
        sigma = 30
    )

    return img6
```

```python
import os
import re
import random
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# DCA1 dataset (Database X-ray Coronary Angiograms)
# 134 angiograms (300x300) + cardiologist ground truth:  N.pgm  /  N_gt.pgm
# ==========================================================
DCA1_ROOT_CANDIDATES = [
    "/kaggle/input/datasets/bard2024/database-x-ray-coronary-angiograms-dca1",
    "/kaggle/input/database-x-ray-coronary-angiograms-dca1",
]
DCA1_ROOT = next((p for p in DCA1_ROOT_CANDIDATES if os.path.isdir(p)), "/kaggle/input")
IMG_EXTS = (".pgm", ".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")


def find_dca1_pairs(root):
    """Recursively pair every  N.<ext>  with its  N_gt.<ext>  under root."""
    images, gts = {}, {}
    for dirpath, _, files in os.walk(root):
        for f in files:
            stem, ext = os.path.splitext(f)
            if ext.lower() not in IMG_EXTS:
                continue
            m = re.fullmatch(r"(\d+)_gt", stem, flags=re.IGNORECASE)
            if m:
                gts[int(m.group(1))] = os.path.join(dirpath, f)
            elif re.fullmatch(r"\d+", stem):
                images[int(stem)] = os.path.join(dirpath, f)
    ids = sorted(set(images) & set(gts))
    return [(i, images[i], gts[i]) for i in ids]


dca1_pairs = find_dca1_pairs(DCA1_ROOT)
assert len(dca1_pairs) > 0, f"No DCA1 image/_gt pairs found under {DCA1_ROOT}. Add the DCA1 dataset as notebook input."
print(f"DCA1 root: {DCA1_ROOT}")
print(f"Found {len(dca1_pairs)} image / ground-truth pairs")

# ----------------------------------------------------------
# Split: standard DCA1 protocol -> images 1-100 train, 101-134 test.
# Validation = 20% of the 100 training images (fixed seed).
# ----------------------------------------------------------
n_trainval = 100 if len(dca1_pairs) == 134 else int(round(len(dca1_pairs) * 100 / 134))
trainval_pairs = dca1_pairs[:n_trainval]
test_pairs = dca1_pairs[n_trainval:]

rng = random.Random(42)
idx = list(range(len(trainval_pairs)))
rng.shuffle(idx)
n_val = int(round(0.2 * len(trainval_pairs)))
val_set = set(idx[:n_val])

DCA1_SPLITS = {
    "train": [p for k, p in enumerate(trainval_pairs) if k not in val_set],
    "val":   [p for k, p in enumerate(trainval_pairs) if k in val_set],
    "test":  test_pairs,
}
for s, v in DCA1_SPLITS.items():
    print(f"{s:5s}: {len(v)} images")

# Sample image used by the preprocessing demo cells below
DCA1_SAMPLE_IMAGE = DCA1_SPLITS["train"][0][1]

# ----------------------------------------------------------
# Visualize one sample with its ground-truth overlay
# ----------------------------------------------------------
sample_id, sample_img_path, sample_gt_path = DCA1_SPLITS["test"][0]
img = cv2.imread(sample_img_path, cv2.IMREAD_GRAYSCALE)
gt = cv2.imread(sample_gt_path, cv2.IMREAD_GRAYSCALE)
print("Image shape:", img.shape, "| GT unique values:", np.unique(gt))

masked_labels = np.ma.masked_where(gt == 0, gt)
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(img, cmap="gray"); ax[0].set_title(f"DCA1 image {sample_id}"); ax[0].axis("off")
ax[1].imshow(img, cmap="gray")
ax[1].imshow(masked_labels, cmap="autumn", alpha=0.5)
ax[1].set_title("Ground truth overlay"); ax[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np


# ==========================================================
# FSE: Filtering Smoothing Equalization
# (Chen et al., "SFAG-DeepLabv3+", Neurocomputing 2025)
#
# Pipeline:
#     Original
#       -> Gaussian smoothing filter   (denoise, Eq. 1-2 in paper)
#       -> CDF-based histogram equalization (contrast, Eq. 3-4 in paper)
# ==========================================================

def _cdf_histogram_equalization(img):
    """
    CDF-based global histogram equalization, matching the paper's
    H(a) = round( (CDF(a) - CDF(min)) / (M*N - CDF(min)) * (L-1) )
    where CDF is the cumulative pixel COUNT (not probability).

    Implemented manually for transparency / fidelity to the paper.
    Mathematically equivalent to cv2.equalizeHist.
    """
    L = 256
    M_N = img.size  # total pixel count

    # Histogram (raw counts) and CDF
    hist = cv2.calcHist([img], [0], None, [L], [0, L]).flatten()
    cdf = np.cumsum(hist)

    # First non-zero CDF value (CDF(min))
    nonzero = cdf[cdf > 0]
    cdf_min = nonzero.min() if nonzero.size > 0 else 0

    # Mapping table H(a)
    denom = max(M_N - cdf_min, 1)  # avoid div-by-zero
    mapping = np.round((cdf - cdf_min) / denom * (L - 1))
    mapping = np.clip(mapping, 0, L - 1).astype(np.uint8)

    return mapping[img]


def fse_preprocess(img, kernel_size=5, sigma=1.5, use_cv2_equalize=True):
    """
    Apply FSE (Filtering Smoothing Equalization) enhancement to a
    coronary angiography image.

    Parameters
    ----------
    img : np.ndarray
        Input image, grayscale (H,W) or BGR (H,W,3).
    kernel_size : int
        Gaussian smoothing window size (paper uses 5x5).
    sigma : float
        Standard deviation for the Gaussian filter.
    use_cv2_equalize : bool
        If True, use cv2.equalizeHist (fast, C-optimized, numerically
        identical algorithm). If False, use the manual CDF implementation
        above (useful if you want to inspect/modify the mapping).

    Returns
    -------
    np.ndarray
        FSE-enhanced grayscale image (uint8).
    """
    if img is None:
        raise ValueError("fse_preprocess received a None image")

    # Ensure single-channel grayscale input
    if len(img.shape) == 3:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray = img.copy()

    if gray.dtype != np.uint8:
        gray = cv2.normalize(gray, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # ---- Step 1: Gaussian smoothing filter (denoise) ----
    smoothed = cv2.GaussianBlur(
        gray,
        (kernel_size, kernel_size),
        sigmaX=sigma,
        sigmaY=sigma
    )

    # ---- Step 2: Histogram equalization via CDF remapping (contrast) ----
    if use_cv2_equalize:
        equalized = cv2.equalizeHist(smoothed)
    else:
        equalized = _cdf_histogram_equalization(smoothed)

    return equalized


# ==========================================================
# Quick sanity check / visualization
# ==========================================================
if __name__ == "__main__":
    sample_path = DCA1_SAMPLE_IMAGE
    img = cv2.imread(sample_path, cv2.IMREAD_GRAYSCALE)

    fse_out = fse_preprocess(img)

    fig, ax = plt.subplots(1, 2, figsize=(10, 5))
    ax[0].imshow(img, cmap="gray"); ax[0].set_title("Original"); ax[0].axis("off")
    ax[1].imshow(fse_out, cmap="gray"); ax[1].set_title("FSE"); ax[1].axis("off")
    plt.tight_layout()
    plt.show()
```

```python
import os
import shutil
import cv2
import numpy as np
# ==========================================================
# Root directories
# ==========================================================
OUTPUT_ROOT = "/kaggle/working/dca1"
splits = ["train", "val", "test"]
# DCA1 images are 300x300. Resize to 512x512 here so the rest of the pipeline
# (IMG_SIZE, PadIfNeeded in validation augmentation) behaves exactly as before.
TARGET_SIZE = (512, 512)  # (width, height)
# ==========================================================
# Process each split
# ==========================================================
for split in splits:
    print(f"\nProcessing {split}...")
    output_image_dir = os.path.join(OUTPUT_ROOT, split, "images")
    output_mask_dir = os.path.join(OUTPUT_ROOT, split, "masks")
    os.makedirs(output_image_dir, exist_ok=True)
    os.makedirs(output_mask_dir, exist_ok=True)
    n_done = 0
    for image_id, image_path, gt_path in DCA1_SPLITS[split]:
        out_name = f"{image_id}.png"
        # ------------------------------------------------------
        # Read ground truth -> binary mask (0 / 255)
        # ------------------------------------------------------
        gt = cv2.imread(gt_path, cv2.IMREAD_GRAYSCALE)
        if gt is None:
            continue
        thr = 0 if gt.max() <= 1 else 127
        mask = (gt > thr).astype(np.uint8) * 255
        mask = cv2.resize(mask, TARGET_SIZE, interpolation=cv2.INTER_NEAREST)
        # ------------------------------------------------------
        # Read and preprocess image
        # ------------------------------------------------------
        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        if image is None:
            continue
        image = cv2.resize(image, TARGET_SIZE, interpolation=cv2.INTER_LINEAR)
        processed_image = fse_preprocess(image)        # Save processed image
        cv2.imwrite(
            os.path.join(output_image_dir, out_name),
            processed_image
        )
        # Save mask
        cv2.imwrite(
            os.path.join(output_mask_dir, out_name),
            mask
        )
        n_done += 1
    print(f"✓ Finished {split}: {n_done} images")
print("\nDataset creation complete!")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# Single Scale Retinex
# ==========================================================
def single_scale_retinex(img, sigma=30):

    img = img.astype(np.float32) + 1

    blur = cv2.GaussianBlur(img, (0,0), sigma)

    retinex = np.log(img) - np.log(blur + 1)

    retinex = cv2.normalize(
        retinex,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return retinex.astype(np.uint8)


# ==========================================================
# Multi-scale TopHat
# ==========================================================
def multiscale_tophat(img):

    result = img.astype(np.int16)

    for s in [5,9,13]:

        kernel = cv2.getStructuringElement(
            cv2.MORPH_ELLIPSE,
            (s,s)
        )

        white = cv2.morphologyEx(
            img,
            cv2.MORPH_TOPHAT,
            kernel
        )

        black = cv2.morphologyEx(
            img,
            cv2.MORPH_BLACKHAT,
            kernel
        )

        result += white
        result -= black

    result = np.clip(result,0,255)

    return result.astype(np.uint8)


# ==========================================================
# Guided Filter
# ==========================================================
def guided(img):

    return cv2.ximgproc.guidedFilter(
        guide=img,
        src=img,
        radius=7,
        eps=1e-3
    )


from skimage.filters import frangi

def frangi_filter(img):

    vessel = frangi(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)



def gamma_correction(img, gamma=0.8):

    table = np.array([
        ((i / 255.0) ** gamma) * 255
        for i in np.arange(256)
    ]).astype("uint8")

    return cv2.LUT(img, table)



from skimage.filters import sato

def sato_filter(img):

    vessel = sato(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)


from skimage.filters import hessian

def hessian_filter(img):

    vessel = hessian(
        img/255.0,
        sigmas=range(1,8),
        black_ridges=False
    )

    vessel = cv2.normalize(
        vessel,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return vessel.astype(np.uint8)


def edge_preserving(img):

    img_bgr = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

    out = cv2.edgePreservingFilter(
        img_bgr,
        flags=1,
        sigma_s=60,
        sigma_r=0.4
    )

    return cv2.cvtColor(out, cv2.COLOR_BGR2GRAY)


# ==========================================================
# Complete Pipeline
# ==========================================================
def preprocess(img):

    # CLAHE
    clahe = cv2.createCLAHE(
        clipLimit=2,
        tileGridSize=(8,8)
    )

    img1 = clahe.apply(img)

    # Circular Gaussian
    img2 = cv2.GaussianBlur(
        img1,
        (7,7),
        1.5
    )

    # Guided Filter
    img3 = guided(img2)

    # Multi-scale TopHat
    img4 = multiscale_tophat(img3)

    # edge preserving
    img5 = edge_preserving(
        img4,
    )

    # Retinex instead of Homomorphic
    img6 = single_scale_retinex(
        img5,
        sigma = 30
    )

    # FSE (Filtering Smoothing Equalization) — applied independently to the
    # original image, for a fair side-by-side comparison against this pipeline
    img7 = fse_preprocess(img)   # requires fse_preprocess defined in an earlier cell

    return img1, img2, img3, img4, img5, img6, img7


# ==========================================================
# Read image
# ==========================================================
img = cv2.imread(
    DCA1_SAMPLE_IMAGE,
    cv2.IMREAD_GRAYSCALE
)

clahe, gaussian, guided_img, tophat, edge, retinex, fse = preprocess(img)

# ==========================================================
# Display
# ==========================================================
titles = [
    "Original",
    "CLAHE",
    "Gaussian",
    "Guided",
    "TopHat",
    "Edge",
    "Retinex",
    "FSE"
]

images = [
    img,
    clahe,
    gaussian,
    guided_img,
    tophat,
    edge,
    retinex,
    fse
]

plt.figure(figsize=(28,5))

for i in range(len(images)):
    plt.subplot(1,8,i+1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()

# ==========================================================
# Quick quantitative comparison (std dev, as used in SFAG paper)
# ==========================================================
print("Standard deviation of each stage (higher ≈ more contrast/detail spread):")
for name, im in zip(titles, images):
    print(f"  {name:10s}: {np.std(im):.3f}")
```

```python
###################################################
# Copyright (c) 2019                              #
# Authors: @iArunava <iarunavaofficial@gmail.com> #
#          @AvivSham <mista2311@gmail.com>        #
#                                                 #
# License: BSD License 3.0                        #
#                                                 #
# The Code in this file is distributed for free   #
# usage and modification with proper linkage back #
# to this repository.                             #
###################################################

import torch
import torch.nn as nn

class InitialBlock(nn.Module):
    def __init__ (self,in_channels = 3,out_channels = 13):
        super().__init__()


        self.maxpool = nn.MaxPool2d(kernel_size=2, 
                                      stride = 2, 
                                      padding = 0)

        self.conv = nn.Conv2d(in_channels, 
                                out_channels,
                                kernel_size = 3,
                                stride = 2, 
                                padding = 1)

        self.prelu = nn.PReLU(16)

        self.batchnorm = nn.BatchNorm2d(out_channels)
  
    def forward(self, x):
        
        main = self.conv(x)
        main = self.batchnorm(main)
        
        side = self.maxpool(x)
        
        x = torch.cat((main, side), dim=1)
        x = self.prelu(x)
        
        return x
```

```python
###################################################
# Copyright (c) 2019                              #
# Authors: @iArunava <iarunavaofficial@gmail.com> #
#          @AvivSham <mista2311@gmail.com>        #
#                                                 #
# License: BSD License 3.0                        #
#                                                 #
# The Code in this file is distributed for free   #
# usage and modification with proper linkage back #
# to this repository.                             #
###################################################

import torch
import torch.nn as nn


class RDDNeck(nn.Module):
    def __init__(self, dilation, in_channels, out_channels, down_flag, relu=False, projection_ratio=4, p=0.1):
        
        super().__init__()
        
        # Define class variables
        self.in_channels = in_channels
        
        self.out_channels = out_channels
        self.dilation = dilation
        self.down_flag = down_flag

        if down_flag:
            self.stride = 2
            self.reduced_depth = int(in_channels // projection_ratio)
        else:
            self.stride = 1
            self.reduced_depth = int(out_channels // projection_ratio)
        
        if relu:
            activation = nn.ReLU()
        else:
            activation = nn.PReLU()
        
        self.maxpool = nn.MaxPool2d(kernel_size = 2,
                                      stride = 2,
                                      padding = 0, return_indices=True)
        

        
        self.dropout = nn.Dropout2d(p=p)

        self.conv1 = nn.Conv2d(in_channels = self.in_channels,
                               out_channels = self.reduced_depth,
                               kernel_size = 1,
                               stride = 1,
                               padding = 0,
                               bias = False,
                               dilation = 1)
        
        self.prelu1 = activation
        
        self.conv2 = nn.Conv2d(in_channels = self.reduced_depth,
                                  out_channels = self.reduced_depth,
                                  kernel_size = 3,
                                  stride = self.stride,
                                  padding = self.dilation,
                                  bias = True,
                                  dilation = self.dilation)
                                  
        self.prelu2 = activation
        
        self.conv3 = nn.Conv2d(in_channels = self.reduced_depth,
                                  out_channels = self.out_channels,
                                  kernel_size = 1,
                                  stride = 1,
                                  padding = 0,
                                  bias = False,
                                  dilation = 1)
        
        self.prelu3 = activation
        
        self.batchnorm1 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm2 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm3 = nn.BatchNorm2d(self.out_channels)
        
        
    def forward(self, x):
        
        bs = x.size()[0]
        x_copy = x
        
        # Side Branch
        x = self.conv1(x)
        x = self.batchnorm1(x)
        x = self.prelu1(x)
        
        x = self.conv2(x)
        x = self.batchnorm2(x)
        x = self.prelu2(x)
        
        x = self.conv3(x)
        x = self.batchnorm3(x)
                
        x = self.dropout(x)
        
        # Main Branch
        if self.down_flag:
            x_copy, indices = self.maxpool(x_copy)
          
        if self.in_channels != self.out_channels:
            out_shape = self.out_channels - self.in_channels
            extras = torch.zeros((bs, out_shape, x.shape[2], x.shape[3]))
            if torch.cuda.is_available():
                extras = extras.cuda()
            x_copy = torch.cat((x_copy, extras), dim = 1)

        # Sum of main and side branches
        x = x + x_copy
        x = self.prelu3(x)
        
        if self.down_flag:
            return x, indices
        else:
            return x
```

```python
import torch
import torch.nn as nn

class UBNeck(nn.Module):
    def __init__(self, in_channels, out_channels, relu=False, projection_ratio=4):
        
        super().__init__()
        
        # Define class variables
        self.in_channels = in_channels
        self.reduced_depth = int(in_channels / projection_ratio)
        self.out_channels = out_channels
        
        
        if relu:
            activation = nn.ReLU()
        else:
            activation = nn.PReLU()
        
        self.unpool = nn.MaxUnpool2d(kernel_size = 2,
                                     stride = 2)
        
        self.main_conv = nn.Conv2d(in_channels = self.in_channels,
                                    out_channels = self.out_channels,
                                    kernel_size = 1)
        
        self.dropout = nn.Dropout2d(p=0.1)
        
        self.convt1 = nn.ConvTranspose2d(in_channels = self.in_channels,
                               out_channels = self.reduced_depth,
                               kernel_size = 1,
                               padding = 0,
                               bias = False)
        
        
        self.prelu1 = activation
        
        self.convt2 = nn.ConvTranspose2d(in_channels = self.reduced_depth,
                                  out_channels = self.reduced_depth,
                                  kernel_size = 3,
                                  stride = 2,
                                  padding = 1,
                                  output_padding = 1,
                                  bias = False)
        
        self.prelu2 = activation
        
        self.convt3 = nn.ConvTranspose2d(in_channels = self.reduced_depth,
                                  out_channels = self.out_channels,
                                  kernel_size = 1,
                                  padding = 0,
                                  bias = False)
        
        self.prelu3 = activation
        
        self.batchnorm1 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm2 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm3 = nn.BatchNorm2d(self.out_channels)
        
    def forward(self, x, indices):
        x_copy = x
        
        # Side Branch
        x = self.convt1(x)
        x = self.batchnorm1(x)
        x = self.prelu1(x)
        
        x = self.convt2(x)
        x = self.batchnorm2(x)
        x = self.prelu2(x)
        
        x = self.convt3(x)
        x = self.batchnorm3(x)
        
        x = self.dropout(x)
        
        # Main Branch
        
        x_copy = self.main_conv(x_copy)
        x_copy = self.unpool(x_copy, indices, output_size=x.size())
        
        # Concat
        x = x + x_copy
        x = self.prelu3(x)
        
        return x
```

```python
###################################################
# Copyright (c) 2019                              #
# Authors: @iArunava <iarunavaofficial@gmail.com> #
#          @AvivSham <mista2311@gmail.com>        #
#                                                 #
# License: BSD License 3.0                        #
#                                                 #
# The Code in this file is distributed for free   #
# usage and modification with proper linkage back #
# to this repository.                             #
###################################################

import torch
import torch.nn as nn

class ASNeck(nn.Module):
    def __init__(self, in_channels, out_channels, projection_ratio=4):
        
        super().__init__()
        
        # Define class variables
        self.in_channels = in_channels
        self.reduced_depth = int(in_channels / projection_ratio)
        self.out_channels = out_channels
        
        self.dropout = nn.Dropout2d(p=0.1)
        
        self.conv1 = nn.Conv2d(in_channels = self.in_channels,
                               out_channels = self.reduced_depth,
                               kernel_size = 1,
                               stride = 1,
                               padding = 0,
                               bias = False)
        
        self.prelu1 = nn.PReLU()
        
        self.conv21 = nn.Conv2d(in_channels = self.reduced_depth,
                                  out_channels = self.reduced_depth,
                                  kernel_size = (1, 5),
                                  stride = 1,
                                  padding = (0, 2),
                                  bias = False)
        
        self.conv22 = nn.Conv2d(in_channels = self.reduced_depth,
                                  out_channels = self.reduced_depth,
                                  kernel_size = (5, 1),
                                  stride = 1,
                                  padding = (2, 0),
                                  bias = False)
        
        self.prelu2 = nn.PReLU()
        
        self.conv3 = nn.Conv2d(in_channels = self.reduced_depth,
                                  out_channels = self.out_channels,
                                  kernel_size = 1,
                                  stride = 1,
                                  padding = 0,
                                  bias = False)
        
        self.prelu3 = nn.PReLU()
        
        self.batchnorm1 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm2 = nn.BatchNorm2d(self.reduced_depth)
        self.batchnorm3 = nn.BatchNorm2d(self.out_channels)
        
    def forward(self, x):
        bs = x.size()[0]
        x_copy = x
        
        # Side Branch
        x = self.conv1(x)
        x = self.batchnorm1(x)
        x = self.prelu1(x)
        
        x = self.conv21(x)
        x = self.conv22(x)
        x = self.batchnorm2(x)
        x = self.prelu2(x)
        
        x = self.conv3(x)
                
        x = self.dropout(x)
        x = self.batchnorm3(x)
        
        # Main Branch
        
        if self.in_channels != self.out_channels:
            out_shape = self.out_channels - self.in_channels
            extras = torch.zeros((bs, out_shape, x.shape[2], x.shape[3]))
            if torch.cuda.is_available():
                extras = extras.cuda()
            x_copy = torch.cat((x_copy, extras), dim = 1)
        
        # Sum of main and side branches
        x = x + x_copy
        x = self.prelu3(x)
        
        return x
```

```python
##################################################################
# Reproducing the paper                                          #
# ENet - Real Time Semantic Segmentation                         #
# Paper: https://arxiv.org/pdf/1606.02147.pdf                    #
#                                                                #
# Copyright (c) 2019                                             #
# Authors: @iArunava <iarunavaofficial@gmail.com>                #
#          @AvivSham <mista2311@gmail.com>                       #
#                                                                #
# License: BSD License 3.0                                       #
#                                                                #
# The Code in this file is distributed for free                  #
# usage and modification with proper credits                     #
# directing back to this repository.                             #
##################################################################

import torch
import torch.nn as nn

class ENet(nn.Module):
    def __init__(self, C):
        super().__init__()
        
        # Define class variables
        self.C = C
        
        # The initial block
        self.init = InitialBlock()
        
        
        # The first bottleneck
        self.b10 = RDDNeck(dilation=1, 
                           in_channels=16, 
                           out_channels=64, 
                           down_flag=True, 
                           p=0.01)
        
        self.b11 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           p=0.01)
        
        self.b12 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           p=0.01)
        
        self.b13 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           p=0.01)
        
        self.b14 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           p=0.01)
        
        
        # The second bottleneck
        self.b20 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=128, 
                           down_flag=True)
        
        self.b21 = RDDNeck(dilation=1, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b22 = RDDNeck(dilation=2, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b23 = ASNeck(in_channels=128, 
                          out_channels=128)
        
        self.b24 = RDDNeck(dilation=4, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b25 = RDDNeck(dilation=1, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b26 = RDDNeck(dilation=8, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b27 = ASNeck(in_channels=128, 
                          out_channels=128)
        
        self.b28 = RDDNeck(dilation=16, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        
        # The third bottleneck
        self.b31 = RDDNeck(dilation=1, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b32 = RDDNeck(dilation=2, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b33 = ASNeck(in_channels=128, 
                          out_channels=128)
        
        self.b34 = RDDNeck(dilation=4, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b35 = RDDNeck(dilation=1, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b36 = RDDNeck(dilation=8, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        self.b37 = ASNeck(in_channels=128, 
                          out_channels=128)
        
        self.b38 = RDDNeck(dilation=16, 
                           in_channels=128, 
                           out_channels=128, 
                           down_flag=False)
        
        
        # The fourth bottleneck
        self.b40 = UBNeck(in_channels=128, 
                          out_channels=64, 
                          relu=True)
        
        self.b41 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           relu=True)
        
        self.b42 = RDDNeck(dilation=1, 
                           in_channels=64, 
                           out_channels=64, 
                           down_flag=False, 
                           relu=True)
        
        
        # The fifth bottleneck
        self.b50 = UBNeck(in_channels=64, 
                          out_channels=16, 
                          relu=True)
        
        self.b51 = RDDNeck(dilation=1, 
                           in_channels=16, 
                           out_channels=16, 
                           down_flag=False, 
                           relu=True)
        
        
        # Final ConvTranspose Layer
        self.fullconv = nn.ConvTranspose2d(in_channels=16, 
                                           out_channels=self.C, 
                                           kernel_size=3, 
                                           stride=2, 
                                           padding=1, 
                                           output_padding=1,
                                           bias=False)

        self.sigmoid = nn.Sigmoid()
        
        
    def forward(self, x):
        
        # The initial block
        x = self.init(x)
        
        # The first bottleneck
        x, i1 = self.b10(x)
        x = self.b11(x)
        x = self.b12(x)
        x = self.b13(x)
        x = self.b14(x)
        
        # The second bottleneck
        x, i2 = self.b20(x)
        x = self.b21(x)
        x = self.b22(x)
        x = self.b23(x)
        x = self.b24(x)
        x = self.b25(x)
        x = self.b26(x)
        x = self.b27(x)
        x = self.b28(x)
        
        # The third bottleneck
        x = self.b31(x)
        x = self.b32(x)
        x = self.b33(x)
        x = self.b34(x)
        x = self.b35(x)
        x = self.b36(x)
        x = self.b37(x)
        x = self.b38(x)
        
        # The fourth bottleneck
        x = self.b40(x, i2)
        x = self.b41(x)
        x = self.b42(x)
        
        # The fifth bottleneck
        x = self.b50(x, i1)
        x = self.b51(x)
        
        # Final ConvTranspose Layer
        x = self.fullconv(x)
        #x = self.sigmoid(x)
        
        return x
```

```python
"""
Lightweight baselines for the EUNet coronary-artery segmentation paper.

    EDANet    [16]  Lo et al., MMAsia 2019
    Fast-SCNN [17]  Poudel et al., BMVC 2019
    UNeXt-S   [29]  Valanarasu & Patel, MICCAI 2022
    MALUNet   [30]  Ruan et al., BIBM 2022

Every model follows the setup in Section III / IV of the paper:
  * input : (B, 3, 512, 512)  - the FSE-enhanced grayscale image copied to 3 channels
  * output: (B, 2, 512, 512)  - raw logits (background, vessel), NO sigmoid/softmax,
            so the same Dice + focal loss (applied per channel with a sigmoid) and the
            0.5 threshold on the vessel channel can be used for every network.

The architectures follow the authors' released code (UNeXt, MALUNet, Fast-SCNN via
the widely used Tramac port) or, for EDANet whose authors released only the layer
table, the layer table itself. Changes from the reference code, all made so the
models plug into one training script:
  * no timm / mmcv dependency;
  * MALUNet returns logits instead of torch.sigmoid(out);
  * Fast-SCNN returns one tensor (no auxiliary head), pads its first conv
    (3x3, stride 2, padding 1) and resizes with explicit target sizes;
  * UNeXt-S uses its own hidden size inside the shifted MLP (identical to the
    original because mlp_ratio = 1).

Dependencies: torch >= 1.10 only.
"""

import math
from typing import List

import torch
import torch.nn as nn
import torch.nn.functional as F

__all__ = ["EDANet", "FastSCNN", "UNeXt_S", "MALUNet", "get_model", "MODELS"]


# =============================================================================
# EDANet [16]
# =============================================================================
class _EDADownsampler(nn.Module):
    """ENet-style downsampler: conv(3x3, s2) || maxpool when C_out > C_in, else conv only."""

    def __init__(self, c_in: int, c_out: int):
        super().__init__()
        self.use_pool = c_in < c_out
        conv_out = c_out - c_in if self.use_pool else c_out
        self.conv = nn.Conv2d(c_in, conv_out, 3, stride=2, padding=1)
        self.pool = nn.MaxPool2d(2, stride=2) if self.use_pool else None
        self.bn = nn.BatchNorm2d(c_out)

    def forward(self, x):
        out = self.conv(x)
        if self.use_pool:
            out = torch.cat([out, self.pool(x)], dim=1)
        return F.relu(self.bn(out))


class _EDAModule(nn.Module):
    """1x1 conv -> (3x1, 1x3) -> dilated (3x1, 1x3), output concatenated to the input."""

    def __init__(self, c_in: int, dilation: int, k: int = 40, drop: float = 0.02):
        super().__init__()
        d = dilation
        self.conv1x1 = nn.Conv2d(c_in, k, 1)
        self.bn0 = nn.BatchNorm2d(k)
        self.conv3x1_1 = nn.Conv2d(k, k, (3, 1), padding=(1, 0))
        self.conv1x3_1 = nn.Conv2d(k, k, (1, 3), padding=(0, 1))
        self.bn1 = nn.BatchNorm2d(k)
        self.conv3x1_2 = nn.Conv2d(k, k, (3, 1), padding=(d, 0), dilation=(d, 1))
        self.conv1x3_2 = nn.Conv2d(k, k, (1, 3), padding=(0, d), dilation=(1, d))
        self.bn2 = nn.BatchNorm2d(k)
        self.dropout = nn.Dropout2d(drop)

    def forward(self, x):
        out = F.relu(self.bn0(self.conv1x1(x)))
        out = F.relu(self.bn1(self.conv1x3_1(self.conv3x1_1(out))))
        out = F.relu(self.bn2(self.conv1x3_2(self.conv3x1_2(out))))
        out = self.dropout(out)
        return torch.cat([out, x], dim=1)


class EDANet(nn.Module):
    """EDANet (0.68 M params with 19 classes). Output stride 8, bilinear x8 upsampling."""

    def __init__(self, in_channels: int = 3, num_classes: int = 2, growth: int = 40):
        super().__init__()
        layers: List[nn.Module] = [
            _EDADownsampler(in_channels, 15),  # 1/2
            _EDADownsampler(15, 60),           # 1/4
        ]
        c = 60
        for d in [1, 1, 1, 2, 2]:              # EDA block 1
            layers.append(_EDAModule(c, d, growth))
            c += growth                        # -> 260
        layers.append(_EDADownsampler(c, 130))  # 1/8 (C_in > C_out: conv only)
        c = 130
        for d in [2, 2, 4, 4, 8, 8, 16, 16]:   # EDA block 2
            layers.append(_EDAModule(c, d, growth))
            c += growth                        # -> 450
        self.features = nn.Sequential(*layers)
        self.project = nn.Conv2d(c, num_classes, 1)
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.normal_(m.weight, 0.0, 0.02)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.normal_(m.weight, 1.0, 0.02)
                nn.init.zeros_(m.bias)

    def forward(self, x):
        size = x.shape[2:]
        out = self.project(self.features(x))
        return F.interpolate(out, size=size, mode="bilinear", align_corners=True)


# =============================================================================
# Fast-SCNN [17]
# =============================================================================
class _ConvBNReLU(nn.Sequential):
    def __init__(self, c_in, c_out, k=3, stride=1, padding=0):
        super().__init__(
            nn.Conv2d(c_in, c_out, k, stride, padding, bias=False),
            nn.BatchNorm2d(c_out),
            nn.ReLU(inplace=True),
        )


class _DSConv(nn.Sequential):
    """Depthwise-separable conv: dw3x3-BN-ReLU-pw1x1-BN-ReLU."""

    def __init__(self, c_in, c_out, stride=1):
        super().__init__(
            nn.Conv2d(c_in, c_in, 3, stride, 1, groups=c_in, bias=False),
            nn.BatchNorm2d(c_in),
            nn.ReLU(inplace=True),
            nn.Conv2d(c_in, c_out, 1, bias=False),
            nn.BatchNorm2d(c_out),
            nn.ReLU(inplace=True),
        )


class _DWConv(nn.Sequential):
    def __init__(self, c_in, c_out, stride=1):
        super().__init__(
            nn.Conv2d(c_in, c_out, 3, stride, 1, groups=c_in, bias=False),
            nn.BatchNorm2d(c_out),
            nn.ReLU(inplace=True),
        )


class _LinearBottleneck(nn.Module):
    """MobileNetV2 inverted residual block."""

    def __init__(self, c_in, c_out, t=6, stride=2):
        super().__init__()
        self.use_shortcut = stride == 1 and c_in == c_out
        self.block = nn.Sequential(
            _ConvBNReLU(c_in, c_in * t, 1),
            _DWConv(c_in * t, c_in * t, stride),
            nn.Conv2d(c_in * t, c_out, 1, bias=False),
            nn.BatchNorm2d(c_out),
        )

    def forward(self, x):
        out = self.block(x)
        return x + out if self.use_shortcut else out


class _PyramidPooling(nn.Module):
    def __init__(self, c_in, c_out, bins=(1, 2, 3, 6)):
        super().__init__()
        inter = c_in // 4
        self.bins = bins
        self.stages = nn.ModuleList([_ConvBNReLU(c_in, inter, 1) for _ in bins])
        self.out = _ConvBNReLU(c_in * 2, c_out, 1)

    def forward(self, x):
        size = x.shape[2:]
        feats = [x]
        for b, conv in zip(self.bins, self.stages):
            y = conv(F.adaptive_avg_pool2d(x, b))
            feats.append(F.interpolate(y, size=size, mode="bilinear", align_corners=True))
        return self.out(torch.cat(feats, dim=1))


class FastSCNN(nn.Module):
    """Fast-SCNN (1.11 M params with 19 classes). Output stride 8."""

    def __init__(self, in_channels: int = 3, num_classes: int = 2):
        super().__init__()
        # Learning to downsample (1/8)
        self.learning_to_downsample = nn.Sequential(
            _ConvBNReLU(in_channels, 32, 3, stride=2, padding=1),
            _DSConv(32, 48, stride=2),
            _DSConv(48, 64, stride=2),
        )
        # Global feature extractor (1/32)
        def stage(c_in, c_out, n, stride):
            blocks = [_LinearBottleneck(c_in, c_out, 6, stride)]
            blocks += [_LinearBottleneck(c_out, c_out, 6, 1) for _ in range(n - 1)]
            return nn.Sequential(*blocks)

        self.global_feature_extractor = nn.Sequential(
            stage(64, 64, 3, 2),
            stage(64, 96, 3, 2),
            stage(96, 128, 3, 1),
            _PyramidPooling(128, 128),
        )
        # Feature fusion
        self.ffm_dw = _DWConv(128, 128, 1)
        self.ffm_low = nn.Sequential(nn.Conv2d(128, 128, 1), nn.BatchNorm2d(128))
        self.ffm_high = nn.Sequential(nn.Conv2d(64, 128, 1), nn.BatchNorm2d(128))
        # Classifier
        self.classifier = nn.Sequential(
            _DSConv(128, 128),
            _DSConv(128, 128),
            nn.Dropout(0.1),
            nn.Conv2d(128, num_classes, 1),
        )

    def forward(self, x):
        size = x.shape[2:]
        high = self.learning_to_downsample(x)
        low = self.global_feature_extractor(high)
        low = F.interpolate(low, size=high.shape[2:], mode="bilinear", align_corners=True)
        low = self.ffm_low(self.ffm_dw(low))
        out = F.relu(self.ffm_high(high) + low, inplace=True)
        out = self.classifier(out)
        return F.interpolate(out, size=size, mode="bilinear", align_corners=True)


# =============================================================================
# UNeXt-S [29]
# =============================================================================
def _init_vit_style(m):
    if isinstance(m, nn.Linear):
        nn.init.trunc_normal_(m.weight, std=0.02)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.LayerNorm):
        nn.init.ones_(m.weight)
        nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Conv2d):
        fan_out = m.kernel_size[0] * m.kernel_size[1] * m.out_channels // m.groups
        nn.init.normal_(m.weight, 0, math.sqrt(2.0 / fan_out))
        if m.bias is not None:
            nn.init.zeros_(m.bias)


def _shift_tokens(x, H, W, shift_size, pad, dim):
    """Shift channel groups along height (dim=2) or width (dim=3). x: (B, N, C)."""
    B, N, C = x.shape
    xn = x.transpose(1, 2).reshape(B, C, H, W)
    xn = F.pad(xn, (pad, pad, pad, pad), "constant", 0)
    chunks = torch.chunk(xn, shift_size, dim=1)
    shifted = [torch.roll(c, s, dim) for c, s in zip(chunks, range(-pad, pad + 1))]
    xs = torch.cat(shifted, dim=1)
    xs = torch.narrow(xs, 2, pad, H)
    xs = torch.narrow(xs, 3, pad, W)
    return xs.reshape(B, C, H * W).transpose(1, 2)


class _DWConvTokens(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dwconv = nn.Conv2d(dim, dim, 3, 1, 1, groups=dim)

    def forward(self, x, H, W):
        B, N, C = x.shape
        x = x.transpose(1, 2).reshape(B, C, H, W)
        return self.dwconv(x).flatten(2).transpose(1, 2)


class _ShiftMLP(nn.Module):
    """Shift(H) -> FC -> DWConv -> GELU -> Shift(W) -> FC."""

    def __init__(self, dim, hidden=None, shift_size=5):
        super().__init__()
        hidden = hidden or dim
        self.fc1 = nn.Linear(dim, hidden)
        self.dwconv = _DWConvTokens(hidden)
        self.act = nn.GELU()
        self.fc2 = nn.Linear(hidden, dim)
        self.shift_size = shift_size
        self.pad = shift_size // 2

    def forward(self, x, H, W):
        x = _shift_tokens(x, H, W, self.shift_size, self.pad, dim=2)
        x = self.act(self.dwconv(self.fc1(x), H, W))
        x = _shift_tokens(x, H, W, self.shift_size, self.pad, dim=3)
        return self.fc2(x)


class _ShiftedBlock(nn.Module):
    def __init__(self, dim, mlp_ratio=1.0):
        super().__init__()
        self.norm = nn.LayerNorm(dim)
        self.mlp = _ShiftMLP(dim, int(dim * mlp_ratio))

    def forward(self, x, H, W):
        return x + self.mlp(self.norm(x), H, W)


class _OverlapPatchEmbed(nn.Module):
    def __init__(self, c_in, dim, patch=3, stride=2):
        super().__init__()
        self.proj = nn.Conv2d(c_in, dim, patch, stride, patch // 2)
        self.norm = nn.LayerNorm(dim)

    def forward(self, x):
        x = self.proj(x)
        _, _, H, W = x.shape
        return self.norm(x.flatten(2).transpose(1, 2)), H, W


class UNeXt_S(nn.Module):
    """UNeXt-S (the small UNeXt of the official repo, about 0.25 M params).

    3 conv stages (8, 16, 32 ch) + 2 tokenized shifted-MLP stages (64, 128 ch),
    additive skip connections, bilinear upsampling. Output stride 1.
    """

    def __init__(self, in_channels: int = 3, num_classes: int = 2,
                 dims=(8, 16, 32, 64, 128)):
        super().__init__()
        c1, c2, c3, c4, c5 = dims
        # Conv encoder
        self.encoder1 = nn.Conv2d(in_channels, c1, 3, padding=1)
        self.encoder2 = nn.Conv2d(c1, c2, 3, padding=1)
        self.encoder3 = nn.Conv2d(c2, c3, 3, padding=1)
        self.ebn1, self.ebn2, self.ebn3 = nn.BatchNorm2d(c1), nn.BatchNorm2d(c2), nn.BatchNorm2d(c3)
        # Tokenized MLP stages
        self.patch_embed3 = _OverlapPatchEmbed(c3, c4)
        self.patch_embed4 = _OverlapPatchEmbed(c4, c5)
        self.block1 = _ShiftedBlock(c4)
        self.block2 = _ShiftedBlock(c5)
        self.norm3, self.norm4 = nn.LayerNorm(c4), nn.LayerNorm(c5)
        self.dblock1 = _ShiftedBlock(c4)
        self.dblock2 = _ShiftedBlock(c3)
        self.dnorm3, self.dnorm4 = nn.LayerNorm(c4), nn.LayerNorm(c3)
        # Decoder
        self.decoder1 = nn.Conv2d(c5, c4, 3, padding=1)
        self.decoder2 = nn.Conv2d(c4, c3, 3, padding=1)
        self.decoder3 = nn.Conv2d(c3, c2, 3, padding=1)
        self.decoder4 = nn.Conv2d(c2, c1, 3, padding=1)
        self.decoder5 = nn.Conv2d(c1, c1, 3, padding=1)
        self.dbn1, self.dbn2 = nn.BatchNorm2d(c4), nn.BatchNorm2d(c3)
        self.dbn3, self.dbn4 = nn.BatchNorm2d(c2), nn.BatchNorm2d(c1)
        self.final = nn.Conv2d(c1, num_classes, 1)
        self.apply(_init_vit_style)

    @staticmethod
    def _up(x):
        return F.interpolate(x, scale_factor=2, mode="bilinear", align_corners=False)

    @staticmethod
    def _to_map(x, B, H, W):
        return x.reshape(B, H, W, -1).permute(0, 3, 1, 2).contiguous()

    def forward(self, x):
        B = x.shape[0]
        # Conv stages
        t1 = out = F.relu(F.max_pool2d(self.ebn1(self.encoder1(x)), 2))    # 1/2
        t2 = out = F.relu(F.max_pool2d(self.ebn2(self.encoder2(out)), 2))  # 1/4
        t3 = out = F.relu(F.max_pool2d(self.ebn3(self.encoder3(out)), 2))  # 1/8
        # Tokenized MLP stage (1/16)
        out, H, W = self.patch_embed3(out)
        out = self._to_map(self.norm3(self.block1(out, H, W)), B, H, W)
        t4 = out
        # Bottleneck (1/32)
        out, H, W = self.patch_embed4(out)
        out = self._to_map(self.norm4(self.block2(out, H, W)), B, H, W)
        # Decoder
        out = F.relu(self._up(self.dbn1(self.decoder1(out)))) + t4          # 1/16
        _, _, H, W = out.shape
        out = self.dblock1(out.flatten(2).transpose(1, 2), H, W)
        out = self._to_map(self.dnorm3(out), B, H, W)
        out = F.relu(self._up(self.dbn2(self.decoder2(out)))) + t3          # 1/8
        _, _, H, W = out.shape
        out = self.dblock2(out.flatten(2).transpose(1, 2), H, W)
        out = self._to_map(self.dnorm4(out), B, H, W)
        out = F.relu(self._up(self.dbn3(self.decoder3(out)))) + t2          # 1/4
        out = F.relu(self._up(self.dbn4(self.decoder4(out)))) + t1          # 1/2
        out = F.relu(self._up(self.decoder5(out)))                          # 1/1
        return self.final(out)


# =============================================================================
# MALUNet [30]
# =============================================================================
class _DepthWiseConv2d(nn.Module):
    def __init__(self, c_in, c_out, k=3, padding=1, stride=1, dilation=1):
        super().__init__()
        self.conv1 = nn.Conv2d(c_in, c_in, k, stride, padding, dilation, groups=c_in)
        self.norm = nn.GroupNorm(4, c_in)
        self.conv2 = nn.Conv2d(c_in, c_out, 1)

    def forward(self, x):
        return self.conv2(self.norm(self.conv1(x)))


class _GatedAttentionUnit(nn.Module):
    def __init__(self, c_in, c_out, k=3):
        super().__init__()
        self.w1 = nn.Sequential(_DepthWiseConv2d(c_in, c_in, k, k // 2), nn.Sigmoid())
        self.w2 = nn.Sequential(_DepthWiseConv2d(c_in, c_in, k + 2, (k + 2) // 2), nn.GELU())
        self.wo = nn.Sequential(_DepthWiseConv2d(c_in, c_out, k), nn.GELU())
        self.cw = nn.Conv2d(c_in, c_out, 1)

    def forward(self, x):
        return self.wo(self.w1(x) * self.w2(x)) + self.cw(x)


class _DilatedGatedAttention(nn.Module):
    """DGA: 4 channel groups with depthwise dilated convs (7, 5, 2, 1), then a GAU."""

    def __init__(self, c_in, c_out, k=3, dilations=(7, 5, 2, 1)):
        super().__init__()
        g = c_in // 4
        self.mda = nn.ModuleList([
            nn.Conv2d(g, g, k, 1, (k + (k - 1) * (d - 1)) // 2, dilation=d, groups=g)
            for d in dilations
        ])
        self.norm = nn.GroupNorm(4, c_in)
        self.conv = nn.Conv2d(c_in, c_in, 1)
        self.gau = _GatedAttentionUnit(c_in, c_out, 3)

    def forward(self, x):
        xs = torch.chunk(x, 4, dim=1)
        x = torch.cat([conv(xi) for conv, xi in zip(self.mda, xs)], dim=1)
        x = F.gelu(self.conv(self.norm(x)))
        return self.gau(x)


class _EAblock(nn.Module):
    """External attention with double normalization."""

    def __init__(self, c):
        super().__init__()
        self.conv1 = nn.Conv2d(c, c, 1)
        self.k = c * 4
        self.linear_0 = nn.Conv1d(c, self.k, 1, bias=False)
        self.linear_1 = nn.Conv1d(self.k, c, 1, bias=False)
        self.linear_1.weight.data = self.linear_0.weight.data.permute(1, 0, 2).clone()
        self.conv2 = nn.Conv2d(c, c, 1, bias=False)
        self.norm = nn.GroupNorm(4, c)

    def forward(self, x):
        idn = x
        x = self.conv1(x)
        b, c, h, w = x.shape
        attn = self.linear_0(x.view(b, c, h * w))
        attn = F.softmax(attn, dim=-1)
        attn = attn / (1e-9 + attn.sum(dim=1, keepdim=True))
        x = self.linear_1(attn).view(b, c, h, w)
        x = self.norm(self.conv2(x))
        return F.gelu(x + idn)


class _ChannelAttBridge(nn.Module):
    def __init__(self, c_list, split_att="fc"):
        super().__init__()
        c_sum = sum(c_list) - c_list[-1]
        self.split_att = split_att
        self.get_all_att = nn.Conv1d(1, 1, 3, padding=1, bias=False)
        make = (lambda co: nn.Linear(c_sum, co)) if split_att == "fc" else (lambda co: nn.Conv1d(c_sum, co, 1))
        self.atts = nn.ModuleList([make(c) for c in c_list[:5]])

    def forward(self, ts):
        att = torch.cat([F.adaptive_avg_pool2d(t, 1) for t in ts], dim=1)  # B, Csum, 1, 1
        att = self.get_all_att(att.squeeze(-1).transpose(-1, -2))         # B, 1, Csum
        if self.split_att != "fc":
            att = att.transpose(-1, -2)
        out = []
        for fc, t in zip(self.atts, ts):
            a = torch.sigmoid(fc(att))
            a = a.transpose(-1, -2).unsqueeze(-1) if self.split_att == "fc" else a.unsqueeze(-1)
            out.append(a.expand_as(t))
        return out


class _SpatialAttBridge(nn.Module):
    def __init__(self):
        super().__init__()
        self.shared = nn.Sequential(nn.Conv2d(2, 1, 7, padding=9, dilation=3), nn.Sigmoid())

    def forward(self, ts):
        return [self.shared(torch.cat([t.mean(1, keepdim=True), t.amax(1, keepdim=True)], 1))
                for t in ts]


class _SCAttBridge(nn.Module):
    def __init__(self, c_list, split_att="fc"):
        super().__init__()
        self.catt = _ChannelAttBridge(c_list, split_att)
        self.satt = _SpatialAttBridge()

    def forward(self, ts):
        r = ts
        ts = [s * t for s, t in zip(self.satt(ts), ts)]
        r_ = ts
        ts = [t + ri for t, ri in zip(ts, r)]
        ts = [c * t for c, t in zip(self.catt(ts), ts)]
        return [t + ri for t, ri in zip(ts, r_)]


class MALUNet(nn.Module):
    """MALUNet (about 0.18 M params). Channels (8, 16, 24, 32, 48, 64), output stride 1."""

    def __init__(self, in_channels: int = 3, num_classes: int = 2,
                 c_list=(8, 16, 24, 32, 48, 64), split_att: str = "fc", bridge: bool = True):
        super().__init__()
        c = list(c_list)
        self.bridge = bridge
        self.encoder1 = nn.Conv2d(in_channels, c[0], 3, padding=1)
        self.encoder2 = nn.Conv2d(c[0], c[1], 3, padding=1)
        self.encoder3 = nn.Conv2d(c[1], c[2], 3, padding=1)
        self.encoder4 = nn.Sequential(_EAblock(c[2]), _DilatedGatedAttention(c[2], c[3]))
        self.encoder5 = nn.Sequential(_EAblock(c[3]), _DilatedGatedAttention(c[3], c[4]))
        self.encoder6 = nn.Sequential(_EAblock(c[4]), _DilatedGatedAttention(c[4], c[5]))
        if bridge:
            self.scab = _SCAttBridge(c, split_att)
        self.decoder1 = nn.Sequential(_DilatedGatedAttention(c[5], c[4]), _EAblock(c[4]))
        self.decoder2 = nn.Sequential(_DilatedGatedAttention(c[4], c[3]), _EAblock(c[3]))
        self.decoder3 = nn.Sequential(_DilatedGatedAttention(c[3], c[2]), _EAblock(c[2]))
        self.decoder4 = nn.Conv2d(c[2], c[1], 3, padding=1)
        self.decoder5 = nn.Conv2d(c[1], c[0], 3, padding=1)
        self.ebn = nn.ModuleList([nn.GroupNorm(4, c[i]) for i in range(5)])
        self.dbn = nn.ModuleList([nn.GroupNorm(4, c[i]) for i in (4, 3, 2, 1, 0)])
        self.final = nn.Conv2d(c[0], num_classes, 1)
        self.apply(self._init_weights)

    @staticmethod
    def _init_weights(m):
        if isinstance(m, nn.Linear):
            nn.init.trunc_normal_(m.weight, std=0.02)
            if m.bias is not None:
                nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Conv1d):
            n = m.kernel_size[0] * m.out_channels
            nn.init.normal_(m.weight, 0, math.sqrt(2.0 / n))
        elif isinstance(m, nn.Conv2d):
            fan_out = m.kernel_size[0] * m.kernel_size[1] * m.out_channels // m.groups
            nn.init.normal_(m.weight, 0, math.sqrt(2.0 / fan_out))
            if m.bias is not None:
                nn.init.zeros_(m.bias)

    @staticmethod
    def _up(x):
        return F.interpolate(x, scale_factor=2, mode="bilinear", align_corners=True)

    def forward(self, x):
        encs = [self.encoder1, self.encoder2, self.encoder3, self.encoder4, self.encoder5]
        skips = []
        out = x
        for enc, bn in zip(encs, self.ebn):
            out = F.gelu(F.max_pool2d(bn(enc(out)), 2))   # 1/2 ... 1/32
            skips.append(out)
        if self.bridge:
            skips = self.scab(skips)
        t1, t2, t3, t4, t5 = skips

        out = F.gelu(self.encoder6(out))                                    # 1/32
        out = F.gelu(self.dbn[0](self.decoder1(out))) + t5                  # 1/32
        out = F.gelu(self._up(self.dbn[1](self.decoder2(out)))) + t4        # 1/16
        out = F.gelu(self._up(self.dbn[2](self.decoder3(out)))) + t3        # 1/8
        out = F.gelu(self._up(self.dbn[3](self.decoder4(out)))) + t2        # 1/4
        out = F.gelu(self._up(self.dbn[4](self.decoder5(out)))) + t1        # 1/2
        return self._up(self.final(out))                                    # logits, 1/1


# =============================================================================
# Factory and self-test
# =============================================================================
MODELS = {
    "edanet": EDANet,
    "fastscnn": FastSCNN,
    "unext_s": UNeXt_S,
    "malunet": MALUNet,
}


def get_model(name: str, in_channels: int = 3, num_classes: int = 2) -> nn.Module:
    key = name.lower().replace("-", "").replace(" ", "")
    key = {"unexts": "unext_s", "unext_s": "unext_s"}.get(key, key)
    if key not in MODELS:
        raise ValueError(f"Unknown model '{name}'. Choose from {list(MODELS)}")
    return MODELS[key](in_channels=in_channels, num_classes=num_classes)


if __name__ == "__main__":
    import time

    x = torch.randn(1, 3, 512, 512)
    for name, cls in MODELS.items():
        model = cls(in_channels=3, num_classes=2).eval()
        params = sum(p.numel() for p in model.parameters()) / 1e6
        with torch.no_grad():
            y = model(x)
            t0 = time.time()
            for _ in range(3):
                model(x)
            cpu_fps = 3 / (time.time() - t0)
        gflops = "n/a"
        try:
            from fvcore.nn import FlopCountAnalysis
            fca = FlopCountAnalysis(model, x)
            fca.unsupported_ops_warnings(False)
            fca.uncalled_modules_warnings(False)
            gflops = f"{fca.total() / 1e9:.2f}"  # fvcore counts one fused multiply-add as one FLOP
        except ImportError:
            pass
        print(f"{name:10s} out={tuple(y.shape)}  params={params:.3f} M  "
              f"GFLOPs(MAC)={gflops}  CPU FPS={cpu_fps:.1f}")
```

```python
# Metrics definitions (ensure these are defined earlier in your code)
class DiceCoefficient(torch.nn.Module):
    def __init__(self, threshold=0.5):
        super(DiceCoefficient, self).__init__()
        self.threshold = threshold
        self.__name__ = "DiceCoefficient"  # Add the __name__ attribute

    def forward(self, y_pred, y_true):
        # y_pred = torch.sigmoid(y_pred)  # Apply sigmoid if predictions are logits
        y_pred = (y_pred > self.threshold).float()  # Apply threshold

        intersection = (y_true * y_pred).sum(dim=(2, 3))  # Sum over height and width
        union = (y_true + y_pred).sum(dim=(2, 3))

        dice = 2. * intersection / (union + 1e-6)  # Add epsilon to avoid division by zero
        return dice.mean()  # Mean over the batch



class IoU(torch.nn.Module):
    def __init__(self, threshold=0.5, eps=1e-7, activation=None):
        """
        Intersection over Union (IoU) metric similar to SMP's implementation.
        
        Args:
            threshold (float): Threshold for converting probabilities to binary predictions.
            eps (float): Small value to avoid division by zero.
            activation (callable, optional): Activation function to apply to predictions (e.g., torch.sigmoid).
                                             If None, assumes inputs are already probabilities.
        """
        super(IoU, self).__init__()
        self.threshold = threshold
        self.eps = eps
        self.activation = activation
        self.__name__ = "IoU"  # For compatibility with metric logging

    def forward(self, y_pred, y_true):
        
        # Apply activation if provided (e.g., sigmoid for logits)
        if self.activation is not None:
            y_pred = self.activation(y_pred)
        
        # Convert probabilities to binary predictions using threshold
        y_pred = (y_pred > self.threshold).float()
        
        # Ensure inputs are binary and have matching shapes
        y_true = y_true.float()
        assert y_pred.shape == y_true.shape, f"Shape mismatch: y_pred {y_pred.shape}, y_true {y_true.shape}"

        # Compute intersection and union
        intersection = (y_pred * y_true).sum(dim=(2, 3))  # Sum over H and W dimensions
        union = (y_pred + y_true - y_pred * y_true).sum(dim=(2, 3))  # Union = A + B - A∩B
        
        # Compute IoU with epsilon to avoid division by zero
        iou = (intersection + self.eps) / (union + self.eps)
        
        # Return mean IoU over the batch
        return iou.mean()
```

```python
import torch
import torch.nn as nn
import numpy as np
from skimage.segmentation import find_boundaries
from scipy.spatial.distance import directed_hausdorff


# -----------------------------
#      PRECISION
# -----------------------------
class Precision(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "Precision"

    def forward(self, y_pred, y_true):
        # model already outputs sigmoid → directly threshold
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        TP = (y_pred * y_true).sum()
        FP = (y_pred * (1 - y_true)).sum()

        return TP / (TP + FP + 1e-6)


# -----------------------------
#      RECALL
# -----------------------------
class Recall(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "Recall"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        TP = (y_pred * y_true).sum()
        FN = ((1 - y_pred) * y_true).sum()

        return TP / (TP + FN + 1e-6)


# -----------------------------
#      F2 SCORE
# -----------------------------
class F2Score(nn.Module):
    def __init__(self, threshold=0.5, beta=2):
        super().__init__()
        self.threshold = threshold
        self.beta = beta
        self.__name__ = "F2Score"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        TP = (y_pred * y_true).sum()
        FP = (y_pred * (1 - y_true)).sum()
        FN = ((1 - y_pred) * y_true).sum()

        beta2 = self.beta ** 2
        return (1 + beta2) * TP / ((1 + beta2) * TP + beta2 * FN + FP + 1e-6)


# -----------------------------
#      MAE
# -----------------------------
class MAE(nn.Module):
    def __init__(self):
        super().__init__()
        self.__name__ = "MAE"

    def forward(self, y_pred, y_true):
        # DO NOT APPLY SIGMOID — model already outputs prob map
        return torch.mean(torch.abs(y_pred - y_true.float()))


# -----------------------------
#      BF-SCORE
# -----------------------------
class BFScore(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "BFScore"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        bf_scores = []

        for p, t in zip(y_pred, y_true):
            p = p.squeeze().cpu().numpy()
            t = t.squeeze().cpu().numpy()

            pb = find_boundaries(p).astype(np.uint8)
            tb = find_boundaries(t).astype(np.uint8)

            TP = np.logical_and(pb, tb).sum()
            FP = np.logical_and(pb, 1 - tb).sum()
            FN = np.logical_and(1 - pb, tb).sum()

            bf = (2 * TP) / (2 * TP + FP + FN + 1e-6)
            bf_scores.append(bf)

        return torch.tensor(bf_scores).mean()


# -----------------------------
#      HD95 (95th percentile Hausdorff)
# -----------------------------
class HD95(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "HD95"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        distances = []

        for p, t in zip(y_pred, y_true):
            p = p.squeeze().cpu().numpy()
            t = t.squeeze().cpu().numpy()

            p_pts = np.argwhere(p == 1)
            t_pts = np.argwhere(t == 1)

            if len(p_pts) == 0 or len(t_pts) == 0:
                distances.append(0)
                continue

            d1 = directed_hausdorff(p_pts, t_pts)[0]
            d2 = directed_hausdorff(t_pts, p_pts)[0]

            distances.append(max(d1, d2))

        return torch.tensor(np.percentile(distances, 95))



# -----------------------------
#      SENSITIVITY
# -----------------------------
class Sensitivity(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "Sensitivity"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        TP = (y_pred * y_true).sum()
        FN = ((1 - y_pred) * y_true).sum()

        return TP / (TP + FN + 1e-6)


# -----------------------------
#      SPECIFICITY
# -----------------------------
class Specificity(nn.Module):
    def __init__(self, threshold=0.5):
        super().__init__()
        self.threshold = threshold
        self.__name__ = "Specificity"

    def forward(self, y_pred, y_true):
        y_pred = (y_pred > self.threshold).float()
        y_true = y_true.float()

        TN = ((1 - y_pred) * (1 - y_true)).sum()
        FP = (y_pred * (1 - y_true)).sum()

        return TN / (TN + FP + 1e-6)
```

```python
import torch
import torch.nn as nn

class DiceLoss(nn.Module):
    def __init__(self, smooth=1e-6):
        super().__init__()

    def forward(self, y_pred, y_true):
        y_pred = torch.sigmoid(y_pred)

        y_pred = y_pred.view(y_pred.size(0), -1)
        y_true = y_true.view(y_true.size(0), -1)

        intersection = (y_pred * y_true).sum(dim=1)
        dice = (2. * intersection + 1e-6) / (
            y_pred.sum(dim=1) + y_true.sum(dim=1) + 1e-6
        )

        return 1 - dice.mean()
```

```python
class DiceLoss(nn.Module):
    def __init__(self, smooth=1e-6):
        super().__init__()
        self.smooth = smooth

    def forward(self, y_pred, y_true):
        y_pred = torch.sigmoid(y_pred).flatten(2)          # (B, C, H*W): one Dice per class
        y_true = y_true.flatten(2)
        intersection = (y_pred * y_true).sum(2)
        dice = (2. * intersection + self.smooth) / (y_pred.sum(2) + y_true.sum(2) + self.smooth)
        return 1 - dice.mean()
```

```python
class FocalLoss(nn.Module):
    def __init__(self, alpha=0.8, gamma=2):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, y_pred, y_true):
        bce = nn.functional.binary_cross_entropy_with_logits(
            y_pred,
            y_true,
            reduction='none'
        )

        pt = torch.exp(-bce)

        focal = self.alpha * ((1 - pt) ** self.gamma) * bce

        return focal.mean()
```

```python
class DiceFocalLoss(nn.Module):
    def __init__(self, alpha=0.8, gamma=2):
        super().__init__()
        self.dice = DiceLoss()
        self.focal = FocalLoss(alpha, gamma)

    def forward(self, y_pred, y_true):
        return self.dice(y_pred, y_true) + self.focal(y_pred, y_true)
```

```python
class TverskyLoss(nn.Module):
    def __init__(self, alpha=0.3, beta=0.7, smooth=1e-6):
        super().__init__()
        self.alpha = alpha
        self.beta = beta
        self.smooth = smooth

    def forward(self, y_pred, y_true):
        y_pred = torch.sigmoid(y_pred)

        y_pred = y_pred.view(y_pred.size(0), -1)
        y_true = y_true.view(y_true.size(0), -1)

        TP = (y_pred * y_true).sum(dim=1)
        FP = ((1 - y_true) * y_pred).sum(dim=1)
        FN = (y_true * (1 - y_pred)).sum(dim=1)

        tversky = (TP + self.smooth) / (
            TP + self.alpha * FP + self.beta * FN + self.smooth
        )

        return 1 - tversky.mean()
```

```python
class FocalTverskyLoss(nn.Module):
    def __init__(self,
                 alpha=0.3,
                 beta=0.7,
                 gamma=0.75,
                 smooth=1e-6):
        super().__init__()

        self.alpha = alpha
        self.beta = beta
        self.gamma = gamma
        self.smooth = smooth

    def forward(self, y_pred, y_true):
        y_pred = torch.sigmoid(y_pred)

        y_pred = y_pred.view(y_pred.size(0), -1)
        y_true = y_true.view(y_true.size(0), -1)

        TP = (y_pred * y_true).sum(dim=1)
        FP = ((1 - y_true) * y_pred).sum(dim=1)
        FN = (y_true * (1 - y_pred)).sum(dim=1)

        tversky = (TP + self.smooth) / (
            TP + self.alpha * FP + self.beta * FN + self.smooth
        )

        loss = (1 - tversky) ** self.gamma

        return loss.mean()
```

```python
# Early Stopping class
class EarlyStopping:
    def __init__(self, patience=5, min_delta=0.001, metric='val_iou'):
        self.patience = patience  # Number of epochs to wait for improvement
        self.min_delta = min_delta  # Minimum improvement required
        self.metric = metric  # Metric to monitor ('val_iou', 'val_dice', or 'val_loss')
        self.best_score = None
        self.wait = 0
        self.stop_training = False

    def __call__(self, metrics_dict):
        current_score = metrics_dict[self.metric]
        if self.metric == 'val_loss':
            current_score = -current_score  # Lower loss is better, so negate for consistency
        
        if self.best_score is None:
            self.best_score = current_score
        elif current_score < self.best_score + self.min_delta:
            self.wait += 1
            print(f"No improvement in {self.metric}. Wait: {self.wait}/{self.patience}")
            if self.wait >= self.patience:
                self.stop_training = True
                print(f"Early stopping triggered after {self.wait} epochs without improvement!")
        else:
            self.best_score = current_score
            self.wait = 0
```
